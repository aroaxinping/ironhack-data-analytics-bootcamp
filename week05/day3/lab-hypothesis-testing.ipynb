In this challenge, we will be working with pokemon data. The data can be found here:

- https://raw.githubusercontent.com/data-bootcamp-v4/data/main/pokemon.csv

```python
#libraries
import pandas as pd
import scipy.stats as st
import numpy as np
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/pokemon.csv")
df
```

- We posit that Pokemons of type Dragon have, on average, more HP stats than Grass. Choose the propper test and, with 5% significance, comment your findings.

```python
# H0: mean(HP | Dragon) <= mean(HP | Grass)
# H1: mean(HP | Dragon) >  mean(HP | Grass)
# Two independent samples, unknown/unequal variances -> Welch's t-test, one-tailed

dragon_hp = df[df['Type 1'] == 'Dragon']['HP']
grass_hp = df[df['Type 1'] == 'Grass']['HP']

t_stat, p_two_tailed = st.ttest_ind(dragon_hp, grass_hp, equal_var=False)
# one-tailed p-value: halve the two-tailed p, only valid because t_stat > 0
# (i.e. the sample already points in the direction of H1)
p_one_tailed = p_two_tailed / 2 if t_stat > 0 else 1 - p_two_tailed / 2

print(f"Dragon: n={len(dragon_hp)}, mean HP={dragon_hp.mean():.2f}")
print(f"Grass:  n={len(grass_hp)}, mean HP={grass_hp.mean():.2f}")
print(f"t-statistic: {t_stat:.4f}")
print(f"one-tailed p-value: {p_one_tailed:.6f}")
print(f"Reject H0 at 5%? {p_one_tailed < 0.05}")
```

**Finding:** Dragon-type Pokémon average 83.3 HP vs. 67.3 HP for Grass-type — a real gap, and the one-tailed Welch's t-test confirms it's not noise (p ≈ 0.0008, well under 0.05). We reject H0: Dragon-type Pokémon do have significantly higher HP than Grass-type on average.

- We posit that Legendary Pokemons have different stats (HP, Attack, Defense, Sp.Atk, Sp.Def, Speed) when comparing with Non-Legendary. Choose the propper test and, with 5% significance, comment your findings.


```python
# H0: mean(stat | Legendary) == mean(stat | Non-Legendary), for each stat
# H1: mean(stat | Legendary) != mean(stat | Non-Legendary)
# Only 2 groups (Legendary vs not) but 6 separate outcome variables -> one
# two-sided Welch's t-test per stat, not ANOVA (ANOVA is for 3+ groups on
# a single outcome).

stats_cols = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']
legendary = df[df['Legendary']]
non_legendary = df[~df['Legendary']]

results = []
for col in stats_cols:
    t_stat, p_val = st.ttest_ind(legendary[col], non_legendary[col], equal_var=False)
    results.append({
        'stat': col,
        'legendary_mean': legendary[col].mean(),
        'non_legendary_mean': non_legendary[col].mean(),
        't_stat': t_stat,
        'p_value': p_val,
        'significant_5pct': p_val < 0.05,
    })

results_df = pd.DataFrame(results)
results_df
```

**Finding:** All six stats come back significant at 5% (every p-value is far below 0.05, several under 1e-10) — Legendary Pokémon are stronger across the board, not just on one or two stats. `Sp. Atk` shows the largest gap in significance (p ≈ 1.6e-21), `Defense` the smallest of the six (p ≈ 4.8e-11), but even that is overwhelming evidence against H0.

In this challenge, we will be working with california-housing data. The data can be found here:

```python
df2 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/california_housing.csv")
df2.head()
```

**We posit that houses close to either a school or a hospital are more expensive.**

- School coordinates (-118, 34)
- Hospital coordinates (-122, 37)

We consider a house (neighborhood) to be close to a school or hospital if the distance is lower than 0.50.

Hint:
- Write a function to calculate euclidean distance from each house (neighborhood) to the school and to the hospital.
- Divide your dataset into houses close and far from either a hospital or school.
- Choose the propper test and, with 5% significance, comment your findings.
 

```python
def euclidean_distance(lon, lat, ref_lon, ref_lat):
    return np.sqrt((lon - ref_lon) ** 2 + (lat - ref_lat) ** 2)

SCHOOL = (-118, 34)
HOSPITAL = (-122, 37)

df2['dist_school'] = euclidean_distance(df2['longitude'], df2['latitude'], *SCHOOL)
df2['dist_hospital'] = euclidean_distance(df2['longitude'], df2['latitude'], *HOSPITAL)

df2['close'] = (df2['dist_school'] < 0.50) | (df2['dist_hospital'] < 0.50)

close_value = df2[df2['close']]['median_house_value']
far_value = df2[~df2['close']]['median_house_value']

print(f"Close: n={len(close_value)}, mean value=${close_value.mean():,.0f}")
print(f"Far:   n={len(far_value)}, mean value=${far_value.mean():,.0f}")
```

```python
# H0: mean(value | close) <= mean(value | far)
# H1: mean(value | close) >  mean(value | far)
# Two independent samples, unequal variances -> Welch's t-test, one-tailed

t_stat, p_two_tailed = st.ttest_ind(close_value, far_value, equal_var=False)
p_one_tailed = p_two_tailed / 2 if t_stat > 0 else 1 - p_two_tailed / 2

print(f"t-statistic: {t_stat:.4f}")
print(f"one-tailed p-value: {p_one_tailed:.3e}")
print(f"Reject H0 at 5%? {p_one_tailed < 0.05}")
```

**Finding:** Houses close to a school or hospital average **$246,952**, vs. **$180,678** for houses farther away — about 37% more expensive. The one-tailed Welch's t-test rejects H0 overwhelmingly (p ≈ 1.5e-301, nowhere near the 5% threshold). Proximity to a school or hospital is strongly associated with higher median house value in this dataset — though as with any observational data, this is an association, not proof that proximity itself *causes* the price premium (it could just as easily reflect that schools/hospitals tend to sit in already-desirable neighborhoods).
