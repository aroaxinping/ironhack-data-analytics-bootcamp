## Exercise: Managing Customer Orders

Practicamos listas, diccionarios, sets y tuplas resolviendo un pequeño sistema de pedidos.

**1.** Define una lista llamada `products` con: 't-shirt', 'mug', 'hat', 'book', 'keychain'.

*Lista → colección ordenada que recorreremos en secuencia.*

```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
```

**2.** Crea un diccionario vacío llamado `inventory`.

*Dict → asocia cada producto (clave) con su cantidad (valor).*

```python
inventory = {}
```

**3.** Pide al usuario la cantidad disponible de cada producto y guárdala en `inventory`.

*`input()` devuelve string → lo convertimos con `int()`.*

```python
for product in products:
    quantity = int(input(f"Enter the quantity of {product}s available: "))
    inventory[product] = quantity
```

**4.** Crea un set vacío llamado `customer_orders`.

*Set → los pedidos no se repiten; los duplicados se descartan solos.*

```python
customer_orders = set()
```

**5.** Pide al usuario el nombre de tres productos que el cliente quiere pedir y añádelos a `customer_orders`.

```python
for _ in range(3):
    order = input("Enter the name of a product the customer wants to order: ")
    customer_orders.add(order)
```

**6.** Imprime los productos del set `customer_orders`.

```python
print("Products in customer_orders:", customer_orders)
```

**7.** Calcula las estadísticas del pedido y guárdalas en una tupla llamada `order_status`:
- Total de productos pedidos
- Porcentaje de productos pedidos (sobre el total de productos disponibles)

*Tupla → agrupa datos que no van a cambiar (una "foto" del estado).*

```python
total_products_ordered = len(customer_orders)
percentage_ordered = (total_products_ordered / len(products)) * 100
order_status = (total_products_ordered, percentage_ordered)
```

**8.** Imprime las estadísticas del pedido con el formato indicado.

```python
print("Order Statistics:")
print(f"Total Products Ordered: {order_status[0]}")
print(f"Percentage of Products Ordered: {order_status[1]}%")
```

**9.** Actualiza el inventario restando 1 a la cantidad de cada producto pedido.

```python
for product in customer_orders:
    inventory[product] -= 1
```

**10.** Imprime el inventario actualizado, mostrando la cantidad de cada producto en líneas separadas.

```python
for product, quantity in inventory.items():
    print(f"{product}: {quantity}")
```
