**1.** `initialize_inventory(products)` — same idea as the functions lab, but the loop that builds `inventory` is now a dict comprehension.

````python
def initialize_inventory(products):
    inventory = {product: int(input(f"Enter the quantity of {product}s available: ")) for product in products}
    return inventory
````

**2.** `get_customer_orders()` now asks how many orders there'll be up front, then builds the `customer_orders` set with a set comprehension over `range(num_orders)` — no more `while True` + "add another?" loop.

````python
def get_customer_orders():
    num_orders = int(input("Enter the number of customer orders: "))
    customer_orders = {input("Enter the name of a product that a customer wants to order: ") for _ in range(num_orders)}
    return customer_orders
````

**3.** New `calculate_total_price(customer_orders)` — asks for a price per ordered product and adds them up with a generator expression inside `sum()`. Assumes 1 unit per product, as the instructions say.

````python
def calculate_total_price(customer_orders):
    total_price = sum(float(input(f"Enter the price of {product}: ")) for product in customer_orders)
    return total_price
````

**4.** `update_inventory` rewritten as a dict comprehension: decrements the quantity for ordered products, and the `if` filters out any product whose quantity would drop to 0 — so it disappears from the inventory instead of being printed as `0`.

````python
def update_inventory(customer_orders, inventory):
    updated_inventory = {
        product: (quantity - 1 if product in customer_orders else quantity)
        for product, quantity in inventory.items()
        if product not in customer_orders or quantity - 1 > 0
    }
    return updated_inventory
````

**5.** Order statistics and the two `print_*` helpers are unchanged from the functions lab — no comprehension needed, they're already doing one simple job each. Just relabeled "Unique Products Ordered" since `customer_orders` is a set.

````python
def calculate_order_statistics(customer_orders, products):
    total_products_ordered = len(customer_orders)
    percentage_ordered = (total_products_ordered / len(products)) * 100
    return total_products_ordered, percentage_ordered


def print_order_statistics(order_statistics):
    print("\nOrder Statistics:")
    print(f"Total Products Ordered: {order_statistics[0]}")
    print(f"Percentage of Unique Products Ordered: {order_statistics[1]}")


def print_updated_inventory(inventory):
    print("\nUpdated Inventory:")
    for product, quantity in inventory.items():
        print(f"{product}: {quantity}")
````

**6.** Call everything in sequence — same order as the sample output in the instructions: stats first, then the updated inventory, then the total price (which is the only step still needing input at that point).

````python
products = ["t-shirt", "mug", "hat", "book", "keychain"]

inventory = initialize_inventory(products)
customer_orders = get_customer_orders()

order_statistics = calculate_order_statistics(customer_orders, products)
print_order_statistics(order_statistics)

inventory = update_inventory(customer_orders, inventory)
print_updated_inventory(inventory)

total_price = calculate_total_price(customer_orders)
print(f"Total Price: {total_price}")
````
