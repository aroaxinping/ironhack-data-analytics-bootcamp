In the last lab, you were starting an online store that sells various products. To ensure smooth operations, you developed a program that manages customer orders and inventory.

Follow the steps below to complete the exercise:

2. Instead of asking the user to input the name of three products that a customer wants to order, do the following:
    
    a. Prompt the user to enter the name of a product that a customer wants to order.
    
    b. Add the product name to the "customer_orders" set.
    
    c. Ask the user if they want to add another product (yes/no).

**1.** `products` list.

```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
```

**2.** Empty `inventory` dictionary.

```python
inventory = {}
```

**3.** Ask for the quantity of each product with a `for` loop.

```python
for product in products:
    quantity = int(input(f"Enter the quantity of {product}s available: "))
    inventory[product] = quantity
```

**4.** Empty `customer_orders` set.

```python
customer_orders = set()
```

**5.** Instead of a fixed `for _ in range(3)`, use a `while True` loop: ask for a product, add it, then ask "add another? (yes/no)" and stop when the answer isn't yes.

```python
while True:
    order = input("Enter the name of a product the customer wants to order: ")
    customer_orders.add(order)

    another = input("Do you want to add another product? (yes/no): ")
    if another.lower().strip() != "yes":
        break
```

**6.** Print the products in `customer_orders`.

```python
print("Products in customer_orders:", customer_orders)
```

**7.** Order statistics stored in the `order_status` tuple.

```python
total_products_ordered = len(customer_orders)
percentage_ordered = (total_products_ordered / len(products)) * 100
order_status = (total_products_ordered, percentage_ordered)
```

**8.** Print the order statistics.

```python
print("Order Statistics:")
print(f"Total Products Ordered: {order_status[0]}")
print(f"Percentage of Products Ordered: {order_status[1]}%")
```

**9.** Update the inventory, subtracting 1 only for the products that were actually ordered.

```python
for product in customer_orders:
    if product in inventory:
        inventory[product] -= 1
```

**10.** Print the updated inventory.

```python
for product, quantity in inventory.items():
    print(f"{product}: {quantity}")
```
