```python
#import libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
spaceship = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/spaceship_titanic.csv")
spaceship.head()
```

**Check the shape of your data**

```python
spaceship.shape
```

**Check for data types**

```python
spaceship.dtypes
```

**Check for missing values**

```python
nulls = spaceship.isna().sum()
pd.DataFrame({"nulls": nulls, "%": (nulls / len(spaceship) * 100).round(2)}).sort_values("nulls", ascending=False)
```

Careful with the framing here. **No single column** has more than 2.5% nulls, which
is what "such low amount of missing values" refers to. But the nulls sit in *different*
rows across twelve columns, and `dropna()` removes a row if **any** of them is null —
so the 2%-per-column compounds to **24% of the passengers gone** (8,693 → 6,606).

That is a real cost, not a rounding error. It is still a defensible choice for this
lab (the remaining 6,606 rows are plenty, and the target stays balanced), but on a
smaller dataset, or if the missingness were not random, imputing would clearly beat
dropping. Worth checking the actual row count after `dropna()` rather than trusting
the per-column percentages.

There are multiple strategies to handle missing data

- Removing all rows or all columns containing missing data.
- Filling all missing values with a value (mean in continouos or mode in categorical for example).
- Filling all missing values with an algorithm.

For this exercise, because we have such low amount of null values, we will drop rows containing any missing value. 

```python
rows_before = len(spaceship)
spaceship = spaceship.dropna()

print(f"Rows before: {rows_before}")
print(f"Rows after : {len(spaceship)}")
print(f"Dropped    : {rows_before - len(spaceship)} ({(rows_before - len(spaceship)) / rows_before * 100:.1f}%)")
```

**KNN**

K Nearest Neighbors is a distance based algorithm, and requeries all **input data to be numerical.**

Let's only select numerical columns as our features.

`select_dtypes(include=np.number)` keeps only the numerical columns and drops the
identifiers and categoricals (`PassengerId`, `HomePlanet`, `Cabin`, `Destination`,
`Name`, and the booleans `CryoSleep`/`VIP`). What is left is `Age` plus the five
spending columns.

```python
features = spaceship.select_dtypes(include=np.number)
features.head()
```

And also lets define our target.

```python
target = spaceship["Transported"]
target.value_counts(normalize=True).round(3)
```

**Train Test Split**

Now that we have split the data into **features** and **target** variables and imported the **train_test_split** function, split X and y into X_train, X_test, y_train, and y_test. 80% of the data should be in the training set and 20% in the test set.

```python
X_train, X_test, y_train, y_test = train_test_split(
    features, target, test_size=0.20, random_state=0
)

print(f"X_train: {X_train.shape}   X_test: {X_test.shape}")
```

**Model Selection**

In this exercise we will be using **KNN** as our predictive model.

You need to choose between **Classificator** or **Regressor**. Take into consideration target variable to decide.

The target, `Transported`, is a **boolean** — a category, not a number. So this is a
classification problem and the right tool is `KNeighborsClassifier`, not
`KNeighborsRegressor`.

Initialize a KNN instance without setting any hyperparameter.

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier()   # no hyperparameters set -> default n_neighbors = 5
knn
```

Fit the model to your data.

```python
knn.fit(X_train, y_train)
```

Evaluate your model.

Accuracy plus a fuller picture, since a single number hides *which* class the model gets wrong.

```python
from sklearn.metrics import classification_report, confusion_matrix

print(f"Accuracy on train: {knn.score(X_train, y_train):.4f}")
print(f"Accuracy on test : {knn.score(X_test, y_test):.4f}\n")

y_pred = knn.predict(X_test)
print(classification_report(y_test, y_pred))
print("Confusion matrix (rows = real, cols = predicted):")
print(confusion_matrix(y_test, y_pred))
```

**How good is this, really?**

The target is almost perfectly balanced (~50/50), so a coin flip scores ~50% and
accuracy is a fair metric here — no class-imbalance trap. The model lands well above
that, so it has genuinely learned something from the spending columns.

Two things worth noticing before being happy with it:

- **Train accuracy sits about 4 points above test accuracy** (0.81 vs 0.77). That gap is overfitting.
  With K-NN it is partly structural (a training point is one of its own neighbours),
  but the size of the gap says the default K = 5 is on the low side for this data.
- **The features are not scaled**, and KNN is a distance-based algorithm. `FoodCourt`
  and `VRDeck` run into the thousands while `Age` tops out around 80, so distances
  are dominated by the spending columns and `Age` barely participates. A
  `StandardScaler` before the model is the single biggest improvement available
  here — more than tuning K.
